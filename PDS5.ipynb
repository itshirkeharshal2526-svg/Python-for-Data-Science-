5a. Write a python code to demonstrate importing pandas libraries and create data frame object.

5b. Write a python code to show statistical information on given data set

```python
import statistics as st 
print(st.mean([1, 3, 5, 7, 9, 11, 13]))
print(st.median([1, 3, 5, 7, 9, 11, 13]))
print(st.mode([1, 1, 3, -5, 7, -9, 11]))
print(st.stdev([1, 3, 5, 7, 9, 11]))
print(st.variance([1, 3, 5, 7, 9, 11]))
```

5c.Write a python code to create pandas series from dictionaries. 


```python

dictionary = {'Niraj': 50, 'Rudra': 60, 'Vedant': 80}

series = pd.Series(dictionary, index=['Niraj', 'Vedant', 'Rudra'])

print(series)
```
