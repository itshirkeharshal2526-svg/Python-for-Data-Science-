**1. Import a CSV dataset using Pandas.**

```python
import pandas as pd
df = pd.read_csv("Students.csv")

df.head()
```

**2. Find missing values in the dataset.**

```python
df.isnull().sum()
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

**3. Replace missing numerical values with the mean.**

```python
numerical_columns = df.select_dtypes(include=['number']).columns

df[numerical_columns] = df[numerical_columns].fillna(
    df[numerical_columns].mean()
)

df[numerical_columns].isnull().sum()
```

**4. Remove duplicate records.**

```python
print("Duplicate records:", df.duplicated().sum())
df = df.drop_duplicates()
print("Dataset shape:", df.shape)
```

**5. Display the cleaned dataset.**

```python
print(df)
df.head(10)
```
